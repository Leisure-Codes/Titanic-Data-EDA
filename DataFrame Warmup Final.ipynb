#Transforming the Dataset

###Dropping The Irrelevant Column

```python
titanic.isnull().sum()
```

From our prelimnary analysis we can see that the embarked column has two missing value, as it's too low, we can just simply drop those rows.

```python
titanic.shape
```

```python
titanic=titanic.dropna(subset=['Embarked'])
```

```python
titanic.shape
```

We can see that the Column "Cabin" has lots of missing values, also as the cabin column doesnt add relevance to the data we can just go ahead and drop the entire column.

```python
 titanic.drop('Cabin',inplace=True, axis=1)
```

```python
titanic
```

Now the Only Column that we are left with to deal with missing values is "Age". It's an important feature for analysis, so we cant drop it and we cant just drop the missing values as they make up of 1/8Th of the Data. So we Impute the data
We will create two seprate datasets

1} where Age column is Imputed by Mean

2} where Missing Values are dropped.

```python
titanic_age=titanic.dropna()
```

```python
titanic_age
```

```python
from sklearn.impute import SimpleImputer
imr = SimpleImputer(missing_values='NaN', strategy='mean')
imr = imr.fit(titanic)
titanic_imputed = imr.transform(titanic.values)
```

We can see that the we cant impute the data until we convert the non-nummeric data into numerical encoding. So our next step is to Drop the irrelevant columns and Convert Non-Numeric columns into Numerical columns.

###**Columns to Drop**
###**PassengerId** : *We dont need passenger id as it contains distinct vales which are not useful in analysis.*
### **Passenger name** : *We don't need passenger name as we are not focussing on one particular person.*
###**Ticket** : *We don't Need Ticket Number as it Contains Distinct values and won't be Helpful for Training any Model.*
###**Fare** : *It contains Distinct Values and the range of the Fare cannot be Defined, Also Pclass Column gives us a Straightforward idea about Passengers Ticket Price, As for 3Rd Class price is around 7. While for 1St class is more than 30.*

```python
 titanic_imputed =titanic.drop(['PassengerId','Name','Ticket','Fare'], axis=1)
 titanic_age=titanic_age.drop(['PassengerId','Name','Ticket','Fare'], axis=1)
```

```python
titanic_imputed
```

###Encoding categorical data using one-hot encoding.

```python
titanic_imputed = pd.get_dummies(titanic_imputed, columns = ['Sex', 'Embarked'])
titanic_imputed
```

Now as we have got our relevant columns and all the Data is Numerical, we can now Impute the Missing Age Values

```python
from sklearn.impute import SimpleImputer
import numpy as np
imr = SimpleImputer(missing_values=np.nan, strategy='median')
titanic_imputed.isna().sum()
```

```python
imr = imr.fit(titanic_imputed)
titanic_imputed = imr.transform(titanic_imputed.values)
```

```python
titanic_imputed_df = pd.DataFrame(titanic_imputed,columns=["Survived", "Pclass",	 "Age",	"SibSp",	"Parch",	"Sex_female",	"Sex_male","Embarked_C","Embarked_Q","Embarked_S"])
```

###**Hypothesis :**
*In the Movie it's shown that the Women & children were given first preferrence in evacuations while the ship sank, so from that point, there has to be more Female Survivors along with childrens. So lets test it*.

**Let's Start with dataset 1**

```python
Test1=titanic_age.groupby(['Survived'])
Test1.first()
```

```python
Test2=Test1.get_group(1)
Female=Test2['Sex'].value_counts()['female']
Male=Test2['Sex'].value_counts()['male']
print("Number of Female Passengers That Survived : %d"%Female)
print("Number of Male Passengers That Survived : %d"%Male)
```

We can clearly see that our Hypothesis Holds True. There are more Female Survivors than Male.**

Now lets Check, of all male that survived how many where children

```python
Test3=Test2[Test2['Sex']=='male']
Test4=Test3[Test3['Age']<=18.0]
Test4.count()
```

```python
Test1.count()
```

####Conclusion (Dataset 1):
 *Altogther we have 889 records of people of Which we dropped 177 record with missing Age, so we now have 288 Survived and 424 Dead passenger samples. Out of 288 Passengers that survived 195 were female and 93 were male. Out of all the Male passengers that survived 24 were children and 69 Adults. So we can see that our Hypothesis is correct. As the Women and childrens were evacuated early Their Population is much more as compared to male population that survided.*

**Let's Check with Second Dataset**

```python
Test_Imp_1=titanic_imputed_df.groupby(['Survived'])
Test_Imp_1.first()
```

```python
Test_Imp_2=Test_Imp_1.get_group(1)
Female=int(Test_Imp_2['Sex_female'].sum())
Male=int(Test_Imp_2['Sex_male'].sum())
print("Number of Female Passengers That Survived : %d"%Female)
print("Number of Male Passengers That Survived : %d"%Male)
```

*We can clearly see that our Hypothesis Holds True. There are more Female Survivors than Male.*

Now lets Check, of all male that survived how many where children

```python
Test_Imp_3=Test_Imp_2[Test_Imp_2['Sex_male']==1.0]
Test_Imp_4=Test_Imp_3[Test_Imp_3['Age']<=18.0]
Test_Imp_4.count()
```

```python
Test_Imp_1.count()
```

####**Conclusion (Dataset 2):**
Altogther we have 889 records of people so we now have 340 Survived and 549 Dead passenger samples. Out of 340 Passengers that survived 231 were female and 109 were male. Out of all the Male passengers that survived 24 were children and 85 Adults. So we can see that our Hypothesis is correct. As the Women and childrens were evacuated early Their Population is much more as compared to male population that survided.

####Explore creating a new quantized feature for numeric data

We can Try another method where we can create bins for age groups

```python
bins = [0, 10, 18, 65, np.inf]
names = ['Child', 'Teen', 'Adult', 'Old']

titanic_age['AgeRange'] = pd.cut(titanic_age['Age'], bins, labels=names)

print(titanic_age.dtypes)
```

####**Either we can get a categorical value or we can Convert This ordinal column into a numeric datatype**

```python
titanic_age=titanic_age.dropna(subset=['AgeRange'])
```

```python
names = [0, 1, 2, 3]

titanic_age['AgeRange'] = pd.cut(titanic_age['Age'], bins, labels=names)
titanic_age["AgeRange"] = titanic_age["AgeRange"].astype("int64")
print(titanic_age.dtypes)
```

```python
titanic_age
```

**This method Works Well for data Anlysis but would not give us better result when we go for Classification, regression, or clustering Methods. So we can try One Hot Encoding Instead**

```python
bins = [0, 10, 18, 65, np.inf]
names = ['Child', 'Teen', 'Adult', 'Old']

titanic_imputed_df['AgeRange'] = pd.cut(titanic_imputed_df['Age'], bins, labels=names)
```

```python
titanic_imputed_df = pd.get_dummies(titanic_imputed_df, columns = ['AgeRange'])
titanic_imputed_df
```

```python
childrens=len(titanic_imputed_df.loc[(titanic_imputed_df['AgeRange_Child']==True) & (titanic_imputed_df['Survived']==1.0) & (titanic_imputed_df['Sex_male']==1.0)])
women=len(titanic_imputed_df.loc[(titanic_imputed_df['Sex_female']==1.0) & (titanic_imputed_df['Survived']==1.0)])
total_women_child=childrens+women
total_women_child
```
